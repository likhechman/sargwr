```python
import os
import numpy as np
from sargwr import GWR_SL
import libpysal as ps
import geopandas as gp
import pandas as pd
from libpysal.weights import Queen
```

```python
georgia_data = pd.read_csv(ps.examples.get_path('GData_utm.csv'))
georgia_shp = gp.read_file(ps.examples.get_path('G_utm.shp'))

w = Queen.from_dataframe(georgia_shp, use_index=False, silence_warnings=True)
w.transform = 'r'

y = georgia_data['PctBach'].values.reshape((-1,1))
X = georgia_data[['PctFB', 'PctBlack', 'PctRural']].values
coords = list(zip(georgia_data['X'], georgia_data['Y']))

X = (X - X.mean(axis=0)) / X.std(axis=0)
y = y.reshape((-1,1))
y = (y - y.mean(axis=0)) / y.std(axis=0)
```

```python
model = GWR_SL(kernel='bisquare', use_coords=False, verbose=True)
model.fit(coords, y, X, W=w, variable_names=['PctFB', 'PctBlack', 'PctRural'])
```

```python
print(model.summary())
```

```python
model.monte_carlo_test(n_perm=10)
```
